Multivariate regression

```python
kf=pd.read_csv('https://docs.google.com/spreadsheets/d/e/2PACX-1vRuHyoYZOoo_hu-XGx6VnOVmqaKiod5NKw0zdVS3--NRB6aBantlNWu2Dk3vdjS8Ubbmgq1XVB3_ehZ/pub?gid=0&single=true&output=csv')
kf
```

```python
import math
median = math.floor(kf.bedroom.median())
median
```

```python
kf.bedroom=kf.bedroom.fillna(median)
kf
```

```python
rr=linear_model.LinearRegression()
rr.fit(kf[['area','bedroom','age']],kf.price)
```

```python
rr.coef_
```

```python
rr.intercept_
```

```python
rr.predict([[3000,3,20]])
```
